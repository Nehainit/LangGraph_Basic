```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_ollama import ChatOllama
```

```python
# define state
class BMI_state(TypedDict):
    height_m:float
    weight_kg:float
    BMI:float
    CATGORY:str
```

```python
def bmi_calculator(state: BMI_state ) -> BMI_state:
    state['BMI']=state['height_m']*state['weight_kg']
    print(state)
    return state

def category(state: BMI_state ) -> BMI_state:
     if state['BMI']>45:
         state['CATGORY']='unhealthy'
     else:
         state['CATGORY']='healthy'

     return state

         
    
```

```python
# define your graph
graph=StateGraph(BMI_state)

# add node to your graph

graph.add_node('calculate_bmi',bmi_calculator)
graph.add_node('define_category',category)

# add edges to your graph

graph.add_edge(START,'calculate_bmi')
graph.add_edge('calculate_bmi','define_category')
graph.add_edge('define_category',END)


# compile your graph

workflow=graph.compile()

# invoke the graph
# state={height_m:2.0,weight_kg:45.0}

result=workflow.invoke({'height_m':2.0,'weight_kg':45.0})
```

```python
print(result)
```

```python
# define state
class llm_qa_state(TypedDict):
    question: str
    answer:str
```

```python
def llm_qa(state:llm_qa_state)->llm_qa_state:
    llm=ChatOllama(model="llama3.2:latest")
    state['answer']=llm.invoke(state['question'])
    return state
    
    
```

```python
# define graph
llm_graph=StateGraph(llm_qa_state)
# define nodes
llm_graph.add_node('llm_qa',llm_qa)
# define edges
llm_graph.add_edge(START,'llm_qa')
llm_graph.add_edge('llm_qa',END)

# compile graph
llm_workflow=llm_graph.compile()

# invoke graph

initial_state={'question':"what is captial of india"}
print(llm_workflow.invoke(initial_state))
```

```python
class blog_state(TypedDict):
    topic:str
    outline:str
    blog:str
    
```

```python
llm=ChatOllama(model="llama3.2:latest")
```

```python
def generate_outline(state:blog_state)->blog_state:
    prompt=f'generate an outline for {state['topic']}'
    state['outline']=llm.invoke(prompt).content
    return state
    
def generate_blog(state:blog_state)->blog_state:
    prompt=f'generate blog on {state['topic']}  follow {state['outline']}'
    state['blog']=llm.invoke(prompt).content
    return state
```

```python
blog_graph=StateGraph(blog_state)

blog_graph.add_node('outline_generator',generate_outline)
blog_graph.add_node('blog_generator',generate_blog)
blog_graph.add_edge(START,'outline_generator')
blog_graph.add_edge('outline_generator','blog_generator')
blog_graph.add_edge('blog_generator',END)

blog_workflow=blog_graph.compile()


result=blog_workflow.invoke({'topic':'INDIA'})
 
```
